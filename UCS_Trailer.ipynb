Using the directed graph, and path costs generated in Part 1 of the exercise, conduct a Uniform Cost Search of the trailer as it traverses across the highway network from Nairobi to Kisumu. Implement this search as a Python program that demonstrates the end-to-end search problem that demonstrates the traversal as how it was done in class. Eventually, present the optimal path and attached cost to the path taken.

```python
import heapq

def uniform_cost_search(graph, start, goal):
    queue = [(0, start, [start])]
    visited = set()

    while queue:
        (cost, current, path) = heapq.heappop(queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            return round(cost, 1), path

        for neighbor, weight in graph.get(current, {}).items():
            if neighbor not in visited:
                heapq.heappush(queue, (cost + weight, neighbor, path + [neighbor]))

    return float('inf'), []

directed_highway_graph = {
    'Nairobi': {'Naivasha': 92.8, 'Thika': 45.0},
    'Naivasha': {'Nakuru': 69.7},
    'Thika': {'Murang\'a': 46.0},
    'Nakuru': {'Eldoret': 156.0, 'Kericho': 107.0},
    'Murang\'a': {'Nyeri': 63.0, 'Sagana': 33.0},
    'Eldoret': {'Kitale': 71.0},
    'Kericho': {'Kisumu': 83.9},
    'Nyeri': {'Nanyuki': 57.0},
    'Sagana': {'Embu': 37.0},
    'Kitale': {},
    'Kisumu': {},
    'Nanyuki': {},
    'Embu': {}
}

start_node = 'Nairobi'
goal_node = 'Kisumu'

optimal_cost, optimal_path = uniform_cost_search(directed_highway_graph, start_node, goal_node)

print(f"Executing Uniform Cost Search: {start_node} to {goal_node}...\n")

if optimal_path:
    print("Optimal Path:")
    print(" -> ".join(optimal_path))
    print(f"\nTotal Path Cost: {optimal_cost} km")
else:
    print(f"No valid path found between {start_node} and {goal_node}.")
```
